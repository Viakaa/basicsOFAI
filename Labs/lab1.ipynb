7. Визначити функцію, яка приймає на вході масив і транспонує його. Скористайтесь методом reshape. І не забудьте почати з docstrings.

```python
def transpose_distance(distance):
  """
  Takes array as input and transposes it

  Parameters
  ----------
  distance: array that will be transposed

  Returns
  -------
  transposed_distance: transposed products array
  """

  n, m = distance.shape

  transpose_distance = distance.reshape(m, n)

  return transpose_distance

if __name__ == "__main__":
  distance = np.array([[3,5,2,2],[54,654,123,543],[21,324,545,656]])
  transposed_distance = transpose_distance(distance)
  print(transposed_distance)
```
